# union

```python
A = {1,2,3,4,5}
B = {4,5,6,7,8}
C = {8,9,10}
```

```python
A.union(B) # Union of A and B (All elements from both sets. NO DUPLICATES)
```

```python
B|C
```

```python
A.union(B,C)# Union of A, B and C.
```
