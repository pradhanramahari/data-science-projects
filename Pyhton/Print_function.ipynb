# print() with format method

```python
num1=20
num2=30
add=num1+num2
print('the addition of {} and {} is={}'.format(num1,num2,add))
```

```python
num1=20
num2=30
num3=90
num4=80
add=num1+num2+num3+num4
print('the addition of {} and {} and {} and {} is={}'.format(num1,num2,num3,num4,add))
```
